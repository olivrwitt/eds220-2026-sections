## Detecting Messy Values

According to the metadata, the allowed values for sex are

|Value | Definition |
|------|----------|
| M | Male |
| F | Female |
| M? | possible male |


```python
hares['sex'].value_counts()
# look at the values in the 'sex' column
```

### Clean values

```python
hares['simple_sex'] = hares['sex'].str.upper().str.strip(" ")
hares['simple_sex'].value_counts()
```

```python
hares.loc[hares['simple_sex'] == '?', 'simple_sex'] = np.nan
hares['simple_sex'].value_counts(dropna = False)
```

```python
hares['simple_sex'] = hares['simple_sex'].replace({
    "PF" : "F?"
})
hares['simple_sex'].value_counts(dropna = False)
```

### Calculate mean weight

```python
hares_valid = hares[hares['weight'] > 0]
hares_valid.groupby('simple_sex')['weight'].mean()
```

### Collect code, streamline workflow, and explain results

```python
import pandas as pd
import numpy as np
hares = pd.read_csv("data/55_Hare_Data_2012.txt")
hares['simple_sex'] = hares['sex'].str.upper().str.strip(" ")
hares.loc[hares['simple_sex'] == '?', 'simple_sex'] = np.nan
hares['simple_sex'] = hares['simple_sex'].replace({
    "PF" : "F?"
})
hares_valid = hares[hares['weight'] > 0]
hares_valid.groupby('simple_sex')['weight'].mean()
```
